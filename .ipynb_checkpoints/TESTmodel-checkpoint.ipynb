```python
import geopandas as gpd
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
from scipy.spatial.distance import pdist, squareform
from shapely.geometry import Polygon


def compute_irr(cashflows):
    """Exact IRR solver using polynomial roots where x = 1 + r."""
    roots = np.roots(cashflows)
    real_pos = [r.real for r in roots if np.isreal(r) and r.real > 0]
    if not real_pos:
        return np.nan
    best_x = min(real_pos, key=lambda x: abs(x - 1.0))
    return (best_x - 1.0) * 100.0


class AvanaEstateMeshModel:
    """
    Unified Geospatial, Electrical, 24-Hour DC Droop, Sizing Sweet-Spot,
    and Zero-CapEx Lease-to-Own SPV Financial Model.
    """

    def __init__(
        self,
        estate_name="Avana Estate Willstone Homes",
        geojson_path="avana_estate.geojson",
        perimeter_coords=None,
        # Site & solar engineering parameters
        usable_roof_ratio=0.38,
        panel_wp_per_m2=210,
        psh=5.2,
        performance_ratio=0.78,
        daily_hh_load_kwh=12.0,
        evening_load_share=0.65,
        battery_dod=0.85,
        battery_rte=0.90,
        # Shared DC ring-bus electrical parameters
        dc_bus_v=380.0,
        cable_mm2=2.5,
        max_road_tie_m=42.0,
        # Hardware & civil CapEx assumptions (USD)
        usd_to_kes=129.0,
        pv_usd_per_kwp=750.0,
        batt_usd_per_kwh=220.0,
        router_usd_per_hh=450.0,
        dc_cable_usd_per_m=10.0,
        # KPLC & Lease-to-Own SPV commercial terms
        kplc_tariff_kes=32.0,
        retail_discount_pct=0.15,
        fixed_capacity_kes_mo=2500.0,
        shared_anchor_kwh_day=35.0,
        anchor_tariff_kes=25.0,
        lease_term_years=10,
        tail_service_fee_kes_mo=1800.0,
        annual_om_pct=0.015,
        tariff_escalation_pct=0.04,
        project_years=15,
        discount_rate=0.10,
    ):
        self.estate_name = estate_name
        self.geojson_path = geojson_path
        self.perimeter_coords = perimeter_coords

        self.usable_roof_ratio = usable_roof_ratio
        self.panel_wp_per_m2 = panel_wp_per_m2
        self.psh = psh
        self.performance_ratio = performance_ratio
        self.daily_hh_load_kwh = daily_hh_load_kwh
        self.evening_load_share = evening_load_share
        self.battery_dod = battery_dod
        self.battery_rte = battery_rte

        self.dc_bus_v = dc_bus_v
        self.cable_mm2 = cable_mm2
        self.max_road_tie_m = max_road_tie_m
        self.rho_cu = 0.0175

        self.usd_to_kes = usd_to_kes
        self.pv_usd_per_kwp = pv_usd_per_kwp
        self.batt_usd_per_kwh = batt_usd_per_kwh
        self.router_usd_per_hh = router_usd_per_hh
        self.dc_cable_usd_per_m = dc_cable_usd_per_m

        self.kplc_tariff_kes = kplc_tariff_kes
        self.retail_discount_pct = retail_discount_pct
        self.fixed_capacity_kes_mo = fixed_capacity_kes_mo
        self.shared_anchor_kwh_day = shared_anchor_kwh_day
        self.anchor_tariff_kes = anchor_tariff_kes
        self.lease_term_years = lease_term_years
        self.tail_service_fee_kes_mo = tail_service_fee_kes_mo
        self.annual_om_pct = annual_om_pct
        self.tariff_escalation_pct = tariff_escalation_pct
        self.project_years = project_years
        self.discount_rate = discount_rate

        self.houses = None
        self.coords_m = None
        self.dist_matrix = None
        self.mst = None
        self.ring_graph = None
        self.sim_results = None
        self.sweep_df = None
        self.lto_df = None

    # ---------------------------------------------------------
    # 1. GEOSPATIAL & ELECTRICAL RING NETWORK
    # ---------------------------------------------------------
    def load_and_build_network(self):
        raw = gpd.read_file(self.geojson_path)
        buildings = raw[raw.geometry.type.isin(["Polygon", "MultiPolygon"])].copy()
        if "building" in buildings.columns:
            buildings = buildings[buildings["building"].notna()].copy()

        self.houses = buildings.to_crs(buildings.estimate_utm_crs()).reset_index(drop=True)
        self.houses["house_id"] = [f"H{i+1:02d}" for i in range(len(self.houses))]

        centroids = self.houses.geometry.centroid
        self.coords_m = np.column_stack([centroids.x, centroids.y])
        self.dist_matrix = squareform(pdist(self.coords_m))

        nn_dists = [np.delete(self.dist_matrix[i], i).min() for i in range(len(self.houses))]
        self.houses["nearest_neighbor_m"] = np.round(nn_dists, 1)
        self.houses["roof_area_m2"] = self.houses.geometry.area.round(1)
        self.houses["max_roof_pv_kwp"] = (
            self.houses["roof_area_m2"] * self.usable_roof_ratio * self.panel_wp_per_m2 / 1000.0
        ).round(2)

        # Build MST and automatically close cross-road ring ties for N-1 redundancy
        G = nx.from_numpy_array(self.dist_matrix)
        self.mst = nx.minimum_spanning_tree(G, weight="weight")
        self.ring_graph = self.mst.copy()

        leaves = [n for n, d in self.mst.degree() if d == 1]
        used_leaves = set()
        for i in range(len(leaves)):
            for j in range(i + 1, len(leaves)):
                u, v = leaves[i], leaves[j]
                if u in used_leaves or v in used_leaves:
                    continue
                d_uv = self.dist_matrix[u, v]
                if 18.0 <= d_uv <= self.max_road_tie_m and not self.ring_graph.has_edge(u, v):
                    if nx.shortest_path_length(self.mst, u, v, weight="weight") > 80.0:
                        self.ring_graph.add_edge(u, v, weight=d_uv, is_redundancy_tie=True)
                        used_leaves.update([u, v])

        path_lengths = dict(nx.all_pairs_dijkstra_path_length(self.ring_graph, weight="weight"))
        self.houses["avg_ring_dist_m"] = [
            round(float(np.mean(list(path_lengths[i].values()))), 1) for i in range(len(self.houses))
        ]
        return self

    # ---------------------------------------------------------
    # 2. 24-HOUR NODAL DC VOLTAGE DROOP SIMULATION
    # ---------------------------------------------------------
    def simulate_24h(self, pv_kwp_per_hh=3.0, batt_kwh_per_hh=8.5, seed=42):
        np.random.seed(seed)
        N = len(self.houses)
        dt_h = 0.25
        steps = int(24 / dt_h)
        hours = np.arange(steps) * dt_h

        self.houses["active_pv_kwp"] = float(pv_kwp_per_hh)
        self.houses["active_batt_kwh"] = float(batt_kwh_per_hh)
        self.houses["daily_yield_kwh"] = round(pv_kwp_per_hh * self.psh * self.performance_ratio, 1)

        # Nodal conductance matrix G_bus
        G_bus = np.zeros((N, N))
        for u, v, data in self.ring_graph.edges(data=True):
            r_line = (2 * self.rho_cu * data["weight"]) / self.cable_mm2
            g_line = 1.0 / r_line
            G_bus[u, u] += g_line
            G_bus[v, v] += g_line
            G_bus[u, v] -= g_line
            G_bus[v, u] -= g_line

        solar_irr = np.maximum(0, np.sin(np.pi * (hours - 6.25) / 12.25)) ** 1.35
        cloud_dip = 1.0 - 0.18 * np.exp(-((hours - 14.5) / 1.2) ** 2)
        pv_gen_kw = np.outer(np.full(N, pv_kwp_per_hh) * self.performance_ratio, solar_irr * cloud_dip)

        load_kw = np.zeros((N, steps))
        for i in range(N):
            base = np.random.uniform(0.25, 0.55)
            morn = np.random.uniform(0.8, 2.2) * np.exp(-((hours - np.random.uniform(6.5, 8.0)) / 0.9) ** 2)
            day = np.random.uniform(0.2, 1.8) * np.exp(-((hours - np.random.uniform(12.0, 14.5)) / 1.8) ** 2)
            eve = np.random.uniform(1.5, 3.6) * np.exp(-((hours - np.random.uniform(19.0, 21.0)) / 1.4) ** 2)
            load_kw[i] = base + morn + day + eve

        soc_min, soc_max = 0.15, 0.95
        soc = np.random.uniform(0.30, 0.75, size=N)
        if N >= 27:
            soc[13] = 0.22  # H14 low SoC benchmark
            soc[26] = 0.85  # H27 high SoC benchmark

        p_router_max_kw = 4.0
        g_droop = np.full(N, 0.85)

        soc_hist = np.zeros((N, steps))
        v_bus_hist = np.zeros((N, steps))
        p_share_hist = np.zeros((N, steps))
        p_grid_backup_hist = np.zeros((N, steps))
        line_loss_kw_hist = np.zeros(steps)

        for t in range(steps):
            soc_hist[:, t] = soc
            net_local_kw = pv_gen_kw[:, t] - load_kw[:, t]
            max_chg = np.minimum(5.0, ((soc_max - soc) * batt_kwh_per_hh) / (dt_h * 0.95))
            max_dis = np.minimum(5.0, ((soc - soc_min) * batt_kwh_per_hh * 0.95) / dt_h)

            v_ref = np.clip(self.dc_bus_v + 14.0 * (soc - 0.55) + 4.0 * np.tanh(net_local_kw / 3.0), 368.0, 392.0)
            v_nodes = np.linalg.solve(G_bus + np.diag(g_droop), g_droop * v_ref)

            p_bus_kw = np.clip((v_nodes * g_droop * (v_ref - v_nodes)) / 1000.0, -p_router_max_kw, p_router_max_kw)
            p_batt_req = net_local_kw - p_bus_kw

            for i in range(N):
                if p_batt_req[i] >= 0:
                    p_chg = min(p_batt_req[i], max_chg[i])
                    soc[i] += (p_chg * 0.95 * dt_h) / batt_kwh_per_hh
                else:
                    needed = -p_batt_req[i]
                    p_dis = min(needed, max_dis[i])
                    p_grid_backup_hist[i, t] = max(0.0, needed - p_dis)
                    soc[i] -= (p_dis / 0.95 * dt_h) / batt_kwh_per_hh

            soc = np.clip(soc, soc_min, soc_max)
            v_bus_hist[:, t] = v_nodes
            p_share_hist[:, t] = p_bus_kw

            step_loss_w = sum(
                (((v_nodes[u] - v_nodes[v]) / ((2 * self.rho_cu * d["weight"]) / self.cable_mm2)) ** 2)
                * ((2 * self.rho_cu * d["weight"]) / self.cable_mm2)
                for u, v, d in self.ring_graph.edges(data=True)
            )
            line_loss_kw_hist[t] = step_loss_w / 1000.0

        self.sim_results = {
            "pv_kwp_per_hh": pv_kwp_per_hh,
            "batt_kwh_per_hh": batt_kwh_per_hh,
            "hours": hours,
            "dt_h": dt_h,
            "soc_hist": soc_hist,
            "v_bus_hist": v_bus_hist,
            "p_share_hist": p_share_hist,
            "p_grid_backup_hist": p_grid_backup_hist,
            "line_loss_kw_hist": line_loss_kw_hist,
        }
        return self.sim_results

    # ---------------------------------------------------------
    # 3. SIZING SWEET-SPOT & LEASE-TO-OWN SPV OPTIMIZATION
    # ---------------------------------------------------------
    def run_full_optimization(
        self,
        pv_grid=(2.5, 3.0, 3.5, 4.5, 6.0),
        batt_grid=(5.0, 6.5, 8.5, 10.0, 12.0),
    ):
        n_houses = len(self.houses)
        ring_cable_m = sum(d["weight"] for _, _, d in self.ring_graph.edges(data=True))
        energy_tariff_kes = round(self.kplc_tariff_kes * (1 - self.retail_discount_pct), 2)

        records = []
        for pv_kwp in pv_grid:
            for batt_kwh in batt_grid:
                sim = self.simulate_24h(pv_kwp_per_hh=pv_kwp, batt_kwh_per_hh=batt_kwh, seed=42)
                dt_h = sim["dt_h"]

                # CapEx (USD & KES)
                pv_usd = pv_kwp * n_houses * self.pv_usd_per_kwp
                batt_usd = batt_kwh * n_houses * self.batt_usd_per_kwh
                router_usd = n_houses * self.router_usd_per_hh
                cable_usd = ring_cable_m * self.dc_cable_usd_per_m
                capex_usd = pv_usd + batt_usd + router_usd + cable_usd
                capex_kes = capex_usd * self.usd_to_kes

                # Displaced KPLC Energy
                grid_backup_daily = float(np.sum(sim["p_grid_backup_hist"]) * dt_h)
                total_load_daily = self.daily_hh_load_kwh * n_houses
                displaced_daily = max(0.0, total_load_daily - grid_backup_daily)
                annual_res_kwh = displaced_daily * 365.0 * 0.92
                annual_anchor_kwh = self.shared_anchor_kwh_day * 365.0 * 0.92
                self_suff_pct = (displaced_daily / total_load_daily) * 100.0

                # Direct Ownership LCOE (15-yr discounted)
                om_kes_yr = capex_kes * self.annual_om_pct
                disc_kwh = sum(annual_res_kwh / ((1 + self.discount_rate) ** yr) for yr in range(1, self.project_years + 1))
                disc_cost = capex_kes + sum(om_kes_yr / ((1 + self.discount_rate) ** yr) for yr in range(1, self.project_years + 1))
                lcoe_kes = disc_cost / max(1.0, disc_kwh)

                # Zero-CapEx Lease-to-Own SPV Cash Flows
                cashflows = [-capex_kes]
                cum_cf = -capex_kes
                spv_payback = None
                for yr in range(1, self.project_years + 1):
                    esc = (1 + self.tariff_escalation_pct) ** (yr - 1)
                    if yr <= self.lease_term_years:
                        rev = (
                            (annual_res_kwh * energy_tariff_kes * esc)
                            + (n_houses * self.fixed_capacity_kes_mo * 12 * esc)
                            + (annual_anchor_kwh * self.anchor_tariff_kes * esc)
                        )
                    else:
                        rev = (n_houses * self.tail_service_fee_kes_mo * 12 * esc) + (
                            annual_anchor_kwh * self.anchor_tariff_kes * esc * 0.5
                        )
                    net_cf = rev - om_kes_yr
                    cashflows.append(net_cf)
                    prev_cum = cum_cf
                    cum_cf += net_cf
                    if spv_payback is None and cum_cf >= 0:
                        spv_payback = round((yr - 1) + abs(prev_cum) / net_cf, 1)

                spv_irr = compute_irr(cashflows)

                # Homeowner monthly comparison
                avg_hh_kwh_mo = (annual_res_kwh / n_houses) / 12.0
                kplc_equiv_mo = (avg_hh_kwh_mo * self.kplc_tariff_kes) + 1500.0
                spv_bill_mo = (avg_hh_kwh_mo * energy_tariff_kes) + self.fixed_capacity_kes_mo

                records.append({
                    "pv_kwp_per_hh": pv_kwp,
                    "batt_kwh_per_hh": batt_kwh,
                    "spv_capex_usd": round(capex_usd),
                    "capex_per_hh_usd": round(capex_usd / n_houses),
                    "pv_cost_usd": round(pv_usd),
                    "batt_cost_usd": round(batt_usd),
                    "router_cost_usd": round(router_usd),
                    "cable_cost_usd": round(cable_usd),
                    "self_sufficiency_pct": round(self_suff_pct, 1),
                    "lcoe_kes_kwh": round(lcoe_kes, 1),
                    "hh_monthly_bill_kes": round(spv_bill_mo),
                    "hh_net_savings_kes_mo": round(kplc_equiv_mo - spv_bill_mo),
                    "spv_payback_yrs": spv_payback if spv_payback else 99.0,
                    "spv_15yr_irr_pct": round(spv_irr, 1),
                    "cashflows_m_kes": [round(c / 1e6, 2) for c in cashflows],
                })

        self.sweep_df = pd.DataFrame(records)
        # Re-run simulation on the #1 optimal IRR/payback sweet-spot configuration so plots reflect the winner
        best = self.sweep_df.sort_values(["spv_15yr_irr_pct", "spv_payback_yrs"], ascending=[False, True]).iloc[0]
        self.best_config = best
        self.simulate_24h(pv_kwp_per_hh=best["pv_kwp_per_hh"], batt_kwh_per_hh=best["batt_kwh_per_hh"], seed=42)
        return self

    # ---------------------------------------------------------
    # 4. ALL VISUALIZATIONS (CABLING, 24H DROOP, HEATMAPS, SPV)
    # ---------------------------------------------------------
    def plot_all_dashboards(self, save_prefix="avana_master"):
        best = self.best_config
        n_houses = len(self.houses)
        ring_m = sum(d["weight"] for _, _, d in self.ring_graph.edges(data=True))

        # FIGURE 1: Closed-Ring DC Meshgrid Cabling Map
        fig1, ax = plt.subplots(figsize=(11, 9))
        if self.perimeter_coords:
            perim = gpd.GeoSeries([Polygon(self.perimeter_coords)], crs="EPSG:4326").to_crs(self.houses.crs)
            perim.boundary.plot(ax=ax, color="#333333", linestyle=":", linewidth=1.3)
        self.houses.plot(ax=ax, facecolor="#e8eef5", edgecolor="#4a6fe3", linewidth=1.2, alpha=0.9, zorder=2)

        for u, v, data in self.ring_graph.edges(data=True):
            w = data["weight"]
            x1, y1 = self.coords_m[u]
            x2, y2 = self.coords_m[v]
            is_red = data.get("is_redundancy_tie", False)
            is_road = w > 18.0
            color, lw, ls = ("#7570b3", 2.6, "--") if is_red else (("#d95f02", 2.8, "-") if is_road else ("#1b9e77", 2.2, "-"))
            ax.plot([x1, x2], [y1, y2], color=color, linewidth=lw, linestyle=ls, zorder=4)
            ax.text(
                (x1 + x2) / 2, (y1 + y2) / 2, f"{w:.1f}m", fontsize=6.8, ha="center", va="center",
                bbox=dict(boxstyle="round,pad=0.15", facecolor="white", edgecolor=color, alpha=0.88), zorder=6
            )

        ax.scatter(self.coords_m[:, 0], self.coords_m[:, 1], color="#d90429", s=55, zorder=5, edgecolors="white")
        med_y = np.median(self.coords_m[:, 1])
        for idx, row in self.houses.iterrows():
            y_off = 11 if self.coords_m[idx, 1] > med_y else -14
            ax.annotate(
                row["house_id"], (self.coords_m[idx, 0], self.coords_m[idx, 1]),
                fontsize=7.5, fontweight="bold", color="#1d3557", ha="center", va="center",
                xytext=(0, y_off), textcoords="offset points", zorder=7
            )

        handles = [
            mlines.Line2D([], [], color="#1b9e77", lw=2.2, label="Row DC Trunk (Underground)"),
            mlines.Line2D([], [], color="#d95f02", lw=2.8, label="Primary Cross-Road Tie"),
            mlines.Line2D([], [], color="#7570b3", lw=2.6, ls="--", label="Redundant Ring-Closure Tie"),
            mlines.Line2D([], [], color="#d90429", marker="o", ls="None", markersize=7, label="Home Energy Router Node"),
        ]
        ax.legend(handles=handles, loc="upper right", fontsize=8.5)
        ax.set_title(f"1. {self.estate_name} — {int(self.dc_bus_v)}V DC Closed-Ring Layout ({ring_m:.1f} m)", fontsize=12.5, fontweight="bold")
        ax.set_aspect("equal")
        plt.tight_layout()
        plt.savefig(f"{save_prefix}_1_cabling.png", dpi=300)
        plt.show()

        # FIGURE 2: Sweet-Spot Optimization Heatmaps (SPV IRR & Payback Period)
        piv_irr = self.sweep_df.pivot(index="batt_kwh_per_hh", columns="pv_kwp_per_hh", values="spv_15yr_irr_pct")
        piv_pay = self.sweep_df.pivot(index="batt_kwh_per_hh", columns="pv_kwp_per_hh", values="spv_payback_yrs")

        fig2, axes2 = plt.subplots(1, 2, figsize=(13, 5.2))
        im1 = axes2[0].imshow(piv_irr.values, cmap="YlGn", aspect="auto")
        axes2[0].set_title("SPV Lease-to-Own 15-Yr IRR (%) — Higher is Better", fontweight="bold")
        axes2[0].set_xlabel("PV Capacity per House (kWp)")
        axes2[0].set_ylabel("Battery Capacity per House (kWh)")
        axes2[0].set_xticks(range(len(piv_irr.columns)))
        axes2[0].set_xticklabels([f"{c} kWp" for c in piv_irr.columns])
        axes2[0].set_yticks(range(len(piv_irr.index)))
        axes2[0].set_yticklabels([f"{r} kWh" for r in piv_irr.index])
        for i in range(piv_irr.shape[0]):
            for j in range(piv_irr.shape[1]):
                axes2[0].text(j, i, f"{piv_irr.values[i, j]:.1f}%", ha="center", va="center", fontweight="bold", fontsize=9)
        fig2.colorbar(im1, ax=axes2[0], shrink=0.82, label="IRR (%)")

        im2 = axes2[1].imshow(piv_pay.values, cmap="RdYlGn_r", aspect="auto")
        axes2[1].set_title("SPV Capital Payback Period (Years) — Lower is Better", fontweight="bold")
        axes2[1].set_xlabel("PV Capacity per House (kWp)")
        axes2[1].set_ylabel("Battery Capacity per House (kWh)")
        axes2[1].set_xticks(range(len(piv_pay.columns)))
        axes2[1].set_xticklabels([f"{c} kWp" for c in piv_pay.columns])
        axes2[1].set_yticks(range(len(piv_pay.index)))
        axes2[1].set_yticklabels([f"{r} kWh" for r in piv_pay.index])
        for i in range(piv_pay.shape[0]):
            for j in range(piv_pay.shape[1]):
                axes2[1].text(j, i, f"{piv_pay.values[i, j]:.1f} yr", ha="center", va="center", fontweight="bold", fontsize=9)
        fig2.colorbar(im2, ax=axes2[1], shrink=0.82, label="Years")

        plt.suptitle(f"2. Sizing Sweet-Spot Sweep — Winner: {best['pv_kwp_per_hh']} kWp PV + {best['batt_kwh_per_hh']} kWh BESS (IRR = {best['spv_15yr_irr_pct']}%)", fontsize=12.5, fontweight="bold")
        plt.tight_layout()
        plt.savefig(f"{save_prefix}_2_sweetspots.png", dpi=300)
        plt.show()

        # FIGURE 3: Cost Breakdown & 15-Year Cumulative Cash Flow Curve for Optimal Sweet Spot
        fig3, axes3 = plt.subplots(1, 2, figsize=(13, 5.0))

        # Left: CapEx Breakdown Bar
        labels = ["PV + Inverter", "LiFePO4 BESS", "Energy Routers", "380V DC Ring"]
        costs_k_usd = [
            best["pv_cost_usd"] / 1e3,
            best["batt_cost_usd"] / 1e3,
            best["router_cost_usd"] / 1e3,
            best["cable_cost_usd"] / 1e3,
        ]
        colors = ["#2a9d8f", "#e9c46a", "#e76f51", "#264653"]
        bars = axes3[0].bar(labels, costs_k_usd, color=colors, edgecolor="black", alpha=0.9)
        axes3[0].set_title(f"SPV CapEx Breakdown (Total = ${best['spv_capex_usd']:,.0f} | ${best['capex_per_hh_usd']:,.0f}/hh)", fontweight="bold")
        axes3[0].set_ylabel("Capital Cost ($ Thousands USD)")
        axes3[0].grid(axis="y", alpha=0.3)
        for bar, val in zip(bars, costs_k_usd):
            axes3[0].text(bar.get_x() + bar.get_width() / 2, val + 1.2, f"${val:.1f}k", ha="center", fontweight="bold", fontsize=9)

        # Right: 15-Year Cumulative Cash Flow (KES Millions)
        yrs = np.arange(0, self.project_years + 1)
        cum_cf_m = np.cumsum(best["cashflows_m_kes"])
        axes3[1].axhline(0, color="black", lw=1, ls="--")
        axes3[1].axvline(self.lease_term_years, color="orange", lw=1.2, ls=":", label=f"Ownership Transfer (Yr {self.lease_term_years})")
        axes3[1].plot(yrs, cum_cf_m, marker="o", color="#1d3557", lw=2.4, label="Cumulative Net Cash Flow")
        axes3[1].fill_between(yrs, 0, cum_cf_m, where=(cum_cf_m >= 0), color="#2b9348", alpha=0.2)
        axes3[1].fill_between(yrs, 0, cum_cf_m, where=(cum_cf_m < 0), color="#d90429", alpha=0.15)
        axes3[1].set_title(f"15-Yr SPV Cash Flow (Payback = {best['spv_payback_yrs']} yrs | IRR = {best['spv_15yr_irr_pct']}%)", fontweight="bold")
        axes3[1].set_xlabel("Project Year")
        axes3[1].set_ylabel("Cumulative Cash Flow (Million KES)")
        axes3[1].legend(loc="upper left")
        axes3[1].grid(True, alpha=0.3)

        plt.suptitle("3. Lease-to-Own SPV Capital Structure & Cash Recovery Trajectory", fontsize=12.5, fontweight="bold")
        plt.tight_layout()
        plt.savefig(f"{save_prefix}_3_financials.png", dpi=300)
        plt.show()

        # FIGURE 4: 24-Hour DC Droop Power Flow Simulation for Optimal Kit
        res = self.sim_results
        hours = res["hours"]
        fig4, axes4 = plt.subplots(3, 1, figsize=(11, 9.5), sharex=True)
        for i in range(n_houses):
            axes4[0].plot(hours, res["soc_hist"][i] * 100, color="steelblue", alpha=0.22, lw=1)
        axes4[0].plot(hours, res["soc_hist"][13] * 100, color="#d90429", lw=2.3, label="H14 (Started Low: 22%)")
        axes4[0].plot(hours, res["soc_hist"][26] * 100, color="#2b9348", lw=2.3, label="H27 (Started High: 85%)")
        axes4[0].set_ylabel("Battery SoC (%)")
        axes4[0].set_title(f"4. 24-Hour DC Meshgrid Operation at Sweet-Spot Kit ({best['pv_kwp_per_hh']} kWp + {best['batt_kwh_per_hh']} kWh)")
        axes4[0].legend(loc="lower right")
        axes4[0].grid(True, alpha=0.3)

        axes4[1].axhline(0, color="black", lw=0.8)
        axes4[1].plot(hours, res["p_share_hist"][26], color="#2b9348", lw=2, label="H27 (+Export / -Import)")
        axes4[1].plot(hours, res["p_share_hist"][13], color="#d90429", lw=2, label="H14 (+Export / -Import)")
        axes4[1].fill_between(hours, 0, np.sum(np.maximum(0, res["p_share_hist"]), axis=0) / 5.0, color="orange", alpha=0.2, label="Total Estate P2P Flow (÷5 scale, kW)")
        axes4[1].set_ylabel("DC Bus Power (kW)")
        axes4[1].legend(loc="upper right")
        axes4[1].grid(True, alpha=0.3)

        axes4[2].axhline(self.dc_bus_v, color="black", ls="--", lw=1, label=f"Nominal {int(self.dc_bus_v)} V DC")
        axes4[2].fill_between(hours, res["v_bus_hist"].min(axis=0), res["v_bus_hist"].max(axis=0), color="purple", alpha=0.2, label="Min–Max Bus Spread")
        axes4[2].plot(hours, res["v_bus_hist"].mean(axis=0), color="purple", lw=2, label="Mean DC Ring Voltage")
        axes4[2].set_ylabel("DC Bus Voltage (V)")
        axes4[2].set_xlabel("Hour of Day (00:00 – 24:00)")
        axes4[2].legend(loc="upper right")
        axes4[2].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.savefig(f"{save_prefix}_4_sim24h.png", dpi=300)
        plt.show()


# =========================================================
# RUN THE ENTIRE END-TO-END MODEL FOR AVANA ESTATE
# =========================================================
avana_perim = [
    (37.0599077, -1.2736124),
    (37.0601902, -1.2741680),
    (37.0616449, -1.2734583),
    (37.0613554, -1.2728957),
    (37.0599077, -1.2736124),
]

full_model = (
    AvanaEstateMeshModel(
        estate_name="Avana Estate Willstone Homes",
        geojson_path="avana_estate.geojson",
        perimeter_coords=avana_perim,
    )
    .load_and_build_network()
    .run_full_optimization()
)

print("=== 1. TOP 5 SIZING SWEET SPOTS (RANKED BY SPV IRR & PAYBACK) ===")
cols_top = [
    "pv_kwp_per_hh", "batt_kwh_per_hh", "spv_capex_usd", "capex_per_hh_usd",
    "self_sufficiency_pct", "lcoe_kes_kwh", "hh_monthly_bill_kes",
    "hh_net_savings_kes_mo", "spv_payback_yrs", "spv_15yr_irr_pct"
]
display(full_model.sweep_df.sort_values(["spv_15yr_irr_pct", "spv_payback_yrs"], ascending=[False, True])[cols_top].head(5))

print("\n=== 2. PER-HOUSE GEOSPATIAL & ROOF POTENTIAL TABLE (FIRST 10 HOUSES) ===")
display(full_model.houses[["house_id", "roof_area_m2", "max_roof_pv_kwp", "nearest_neighbor_m", "avg_ring_dist_m"]].head(10))

full_model.plot_all_dashboards(save_prefix="avana_master")
```

```python
%%writefile app.py
import streamlit as st
import geopandas as gpd
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.lines as mlines
from scipy.spatial.distance import pdist, squareform
import tempfile
import os

st.set_page_config(page_title="DC Meshgrid & Energy Router Planner", layout="wide")


def compute_irr(cashflows):
    roots = np.roots(cashflows)
    real_pos = [r.real for r in roots if np.isreal(r) and r.real > 0]
    if not real_pos:
        return np.nan
    best_x = min(real_pos, key=lambda x: abs(x - 1.0))
    return (best_x - 1.0) * 100.0


@st.cache_data
def run_estate_simulation(
    geojson_path,
    estate_name,
    pv_kwp,
    batt_kwh,
    dc_bus_v,
    cable_mm2,
    kplc_tariff_kes,
    retail_discount_pct,
    fixed_cap_kes_mo,
    anchor_kwh_day,
    lease_years,
    pv_usd_kwp,
    batt_usd_kwh,
    router_usd_hh,
):
    raw = gpd.read_file(geojson_path)
    buildings = raw[raw.geometry.type.isin(["Polygon", "MultiPolygon"])].copy()
    if "building" in buildings.columns:
        buildings = buildings[buildings["building"].notna()].copy()

    houses = buildings.to_crs(buildings.estimate_utm_crs()).reset_index(drop=True)
    n_houses = len(houses)
    houses["house_id"] = [f"H{i+1:02d}" for i in range(n_houses)]

    centroids = houses.geometry.centroid
    coords_m = np.column_stack([centroids.x, centroids.y])
    dist_matrix = squareform(pdist(coords_m))

    houses["nearest_neighbor_m"] = np.round([np.delete(dist_matrix[i], i).min() for i in range(n_houses)], 1)
    houses["roof_area_m2"] = houses.geometry.area.round(1)

    # Build MST + N-1 closed ring
    G = nx.from_numpy_array(dist_matrix)
    mst = nx.minimum_spanning_tree(G, weight="weight")
    ring_graph = mst.copy()

    leaves = [n for n, d in mst.degree() if d == 1]
    used = set()
    for i in range(len(leaves)):
        for j in range(i + 1, len(leaves)):
            u, v = leaves[i], leaves[j]
            if u in used or v in used:
                continue
            d_uv = dist_matrix[u, v]
            if 18.0 <= d_uv <= 42.0 and not ring_graph.has_edge(u, v):
                if nx.shortest_path_length(mst, u, v, weight="weight") > 80.0:
                    ring_graph.add_edge(u, v, weight=d_uv, is_redundancy_tie=True)
                    used.update([u, v])

    ring_m = sum(d["weight"] for _, _, d in ring_graph.edges(data=True))

    # 24-hour nodal DC droop simulation
    np.random.seed(42)
    rho_cu = 0.0175
    dt_h = 0.25
    steps = 96
    hours = np.arange(steps) * dt_h

    G_bus = np.zeros((n_houses, n_houses))
    for u, v, data in ring_graph.edges(data=True):
        r_line = (2 * rho_cu * data["weight"]) / cable_mm2
        g_line = 1.0 / r_line
        G_bus[u, u] += g_line
        G_bus[v, v] += g_line
        G_bus[u, v] -= g_line
        G_bus[v, u] -= g_line

    solar_irr = np.maximum(0, np.sin(np.pi * (hours - 6.25) / 12.25)) ** 1.35
    cloud_dip = 1.0 - 0.18 * np.exp(-((hours - 14.5) / 1.2) ** 2)
    pv_gen_kw = np.outer(np.full(n_houses, pv_kwp) * 0.78, solar_irr * cloud_dip)

    load_kw = np.zeros((n_houses, steps))
    for i in range(n_houses):
        base = np.random.uniform(0.25, 0.55)
        morn = np.random.uniform(0.8, 2.2) * np.exp(-((hours - np.random.uniform(6.5, 8.0)) / 0.9) ** 2)
        day = np.random.uniform(0.2, 1.8) * np.exp(-((hours - np.random.uniform(12.0, 14.5)) / 1.8) ** 2)
        eve = np.random.uniform(1.5, 3.6) * np.exp(-((hours - np.random.uniform(19.0, 21.0)) / 1.4) ** 2)
        load_kw[i] = base + morn + day + eve

    soc = np.random.uniform(0.30, 0.75, size=n_houses)
    soc[0] = 0.22
    soc[-1] = 0.85

    g_droop = np.full(n_houses, 0.85)
    soc_hist = np.zeros((n_houses, steps))
    v_bus_hist = np.zeros((n_houses, steps))
    p_share_hist = np.zeros((n_houses, steps))
    grid_backup_hist = np.zeros((n_houses, steps))

    for t in range(steps):
        soc_hist[:, t] = soc
        net_kw = pv_gen_kw[:, t] - load_kw[:, t]
        max_chg = np.minimum(5.0, ((0.95 - soc) * batt_kwh) / (dt_h * 0.95))
        max_dis = np.minimum(5.0, ((soc - 0.15) * batt_kwh * 0.95) / dt_h)

        v_ref = np.clip(dc_bus_v + 14.0 * (soc - 0.55) + 4.0 * np.tanh(net_kw / 3.0), dc_bus_v - 12.0, dc_bus_v + 12.0)
        v_nodes = np.linalg.solve(G_bus + np.diag(g_droop), g_droop * v_ref)
        p_bus_kw = np.clip((v_nodes * g_droop * (v_ref - v_nodes)) / 1000.0, -4.0, 4.0)
        p_batt_req = net_kw - p_bus_kw

        for i in range(n_houses):
            if p_batt_req[i] >= 0:
                p_chg = min(p_batt_req[i], max_chg[i])
                soc[i] += (p_chg * 0.95 * dt_h) / batt_kwh
            else:
                needed = -p_batt_req[i]
                p_dis = min(needed, max_dis[i])
                grid_backup_hist[i, t] = max(0.0, needed - p_dis)
                soc[i] -= (p_dis / 0.95 * dt_h) / batt_kwh

        soc = np.clip(soc, 0.15, 0.95)
        v_bus_hist[:, t] = v_nodes
        p_share_hist[:, t] = p_bus_kw

    # Financials (Lease-to-Own SPV)
    usd_to_kes = 129.0
    pv_cost = pv_kwp * n_houses * pv_usd_kwp
    batt_cost = batt_kwh * n_houses * batt_usd_kwh
    router_cost = n_houses * router_usd_hh
    cable_cost = ring_m * 10.0
    capex_usd = pv_cost + batt_cost + router_cost + cable_cost
    capex_kes = capex_usd * usd_to_kes

    grid_backup_daily = float(np.sum(grid_backup_hist) * dt_h)
    total_load_daily = 12.0 * n_houses
    displaced_daily = max(0.0, total_load_daily - grid_backup_daily)
    self_suff_pct = (displaced_daily / total_load_daily) * 100.0

    annual_res_kwh = displaced_daily * 365.0 * 0.92
    annual_anchor_kwh = anchor_kwh_day * 365.0 * 0.92
    retail_tariff = kplc_tariff_kes * (1 - retail_discount_pct / 100.0)

    cashflows = [-capex_kes]
    cum_cf = -capex_kes
    payback_yr = None
    for yr in range(1, 16):
        esc = (1.04) ** (yr - 1)
        om = capex_kes * 0.015
        if yr <= lease_years:
            rev = (annual_res_kwh * retail_tariff * esc) + (n_houses * fixed_cap_kes_mo * 12 * esc) + (annual_anchor_kwh * 25.0 * esc)
        else:
            rev = (n_houses * 1800.0 * 12 * esc) + (annual_anchor_kwh * 25.0 * esc * 0.5)
        net_cf = rev - om
        cashflows.append(net_cf)
        prev = cum_cf
        cum_cf += net_cf
        if payback_yr is None and cum_cf >= 0:
            payback_yr = round((yr - 1) + abs(prev) / net_cf, 1)

    irr_pct = compute_irr(cashflows)
    p2p_shared_kwh = float(np.sum(np.maximum(0, p_share_hist)) * dt_h)

    return {
        "estate_name": estate_name,
        "houses": houses,
        "coords_m": coords_m,
        "ring_graph": ring_graph,
        "ring_m": round(ring_m, 1),
        "n_houses": n_houses,
        "self_suff_pct": round(self_suff_pct, 1),
        "p2p_shared_kwh": round(p2p_shared_kwh, 1),
        "capex_usd": round(capex_usd),
        "capex_per_hh_usd": round(capex_usd / n_houses),
        "cost_breakdown": {"PV + Inverter": pv_cost, "LiFePO4 BESS": batt_cost, "Energy Routers": router_cost, "DC Ring Cable": cable_cost},
        "payback_yr": payback_yr if payback_yr else 15.0,
        "irr_pct": round(irr_pct, 1),
        "cashflows_m_kes": [c / 1e6 for c in cashflows],
        "hours": hours,
        "soc_hist": soc_hist,
        "v_bus_hist": v_bus_hist,
        "p_share_hist": p_share_hist,
    }


# ================= SIDEBAR CONTROLS =================
st.sidebar.header("1. Select or Upload Estate")
preset_estates = {
    "Avana Estate Willstone Homes (27 Homes)": "avana_estate.geojson",
    "The Kongo's Estate (31 Homes)": "The_Kongo`s_Estate.geojson",
}
selected_label = st.sidebar.selectbox("Built-in Estate GeoJSON", list(preset_estates.keys()))
uploaded_file = st.sidebar.file_uploader("Or Upload New Estate (.geojson)", type=["geojson", "json"])

if uploaded_file is not None:
    tmp = tempfile.NamedTemporaryFile(delete=False, suffix=".geojson")
    tmp.write(uploaded_file.read())
    tmp.close()
    active_path = tmp.name
    active_name = uploaded_file.name.replace(".geojson", "").replace("_", " ")
else:
    active_path = preset_estates[selected_label]
    active_name = selected_label.split(" (")[0]

st.sidebar.header("2. Hardware & DC Mesh Sizing")
pv_kwp = st.sidebar.slider("PV per House (kWp)", 2.0, 6.0, 3.0, 0.5)
batt_kwh = st.sidebar.slider("Battery per House (kWh)", 5.0, 15.0, 10.0, 0.5)
dc_bus_v = st.sidebar.selectbox("Shared DC Bus Voltage (V)", [380.0, 48.0], index=0)
cable_mm2 = st.sidebar.selectbox("DC Ring Copper Size (mm²)", [2.5, 4.0, 6.0, 10.0, 50.0, 95.0], index=0)

st.sidebar.header("3. Lease-to-Own SPV Terms")
kplc_tariff = st.sidebar.slider("KPLC Tariff (KES/kWh)", 25.0, 40.0, 32.0, 1.0)
discount_pct = st.sidebar.slider("Resident Tariff Discount (%)", 5, 30, 15, 1)
fixed_fee = st.sidebar.number_input("Monthly Capacity Fee (KES)", 0, 5000, 2500, 250)
anchor_kwh = st.sidebar.slider("Shared Estate Borehole/Light Load (kWh/day)", 0, 80, 35, 5)
lease_yrs = st.sidebar.slider("Lease-to-Own Transfer Term (Years)", 6, 15, 10, 1)

st.sidebar.header("4. Unit CapEx ($ USD)")
pv_usd = st.sidebar.number_input("PV + Inverter ($/kWp)", 500, 1200, 750, 25)
batt_usd = st.sidebar.number_input("LiFePO4 BESS ($/kWh)", 150, 400, 220, 10)
router_usd = st.sidebar.number_input("Energy Router ($/house)", 250, 800, 450, 25)

res = run_estate_simulation(
    active_path, active_name, pv_kwp, batt_kwh, dc_bus_v, cable_mm2,
    kplc_tariff, discount_pct, fixed_fee, anchor_kwh, lease_yrs,
    pv_usd, batt_usd, router_usd
)

# ================= MAIN DASHBOARD =================
st.title(f"⚡ {res['estate_name']} — DC Meshgrid & Energy Router Planner")

c1, c2, c3, c4, c5, c6 = st.columns(6)
c1.metric("Connected Homes", f"{res['n_houses']}")
c2.metric("Closed DC Ring", f"{res['ring_m']} m", f"{round(res['ring_m']/res['n_houses'], 1)} m/hh")
c3.metric("SPV Total CapEx", f"${res['capex_usd']:,}", f"${res['capex_per_hh_usd']:,}/hh")
c4.metric("Self-Sufficiency", f"{res['self_suff_pct']}%", f"{res['p2p_shared_kwh']} kWh/d P2P")
c5.metric("SPV Payback", f"{res['payback_yr']} yrs")
c6.metric("SPV 15-Yr IRR", f"{res['irr_pct']}%")

tab1, tab2, tab3, tab4 = st.tabs([
    "🗺️ DC Ring Cabling Schematic",
    "🔋 24-Hr DC Droop Simulation",
    "💰 SPV CapEx & Cash Flows",
    "🏢 Multi-Estate Portfolio Roll-Up",
])

with tab1:
    fig1, ax = plt.subplots(figsize=(10, 7.5))
    res["houses"].plot(ax=ax, facecolor="#e8eef5", edgecolor="#4a6fe3", linewidth=1.2)
    coords = res["coords_m"]
    for u, v, d in res["ring_graph"].edges(data=True):
        w = d["weight"]
        is_red = d.get("is_redundancy_tie", False)
        is_road = w > 18.0
        color, lw, ls = ("#7570b3", 2.5, "--") if is_red else (("#d95f02", 2.6, "-") if is_road else ("#1b9e77", 2.0, "-"))
        ax.plot([coords[u, 0], coords[v, 0]], [coords[u, 1], coords[v, 1]], color=color, lw=lw, ls=ls)
        ax.text((coords[u, 0] + coords[v, 0]) / 2, (coords[u, 1] + coords[v, 1]) / 2, f"{w:.1f}m", fontsize=6.5, ha="center",
                bbox=dict(boxstyle="round,pad=0.12", facecolor="white", edgecolor=color, alpha=0.85))
    ax.scatter(coords[:, 0], coords[:, 1], color="#d90429", s=45, zorder=5)
    for idx, row in res["houses"].iterrows():
        ax.annotate(row["house_id"], (coords[idx, 0], coords[idx, 1]), fontsize=7, fontweight="bold", xytext=(0, 6), textcoords="offset points", ha="center")
    ax.set_title(f"{res['estate_name']} — {int(dc_bus_v)}V DC Closed-Ring Topology ({res['ring_m']} m)")
    ax.set_aspect("equal")
    st.pyplot(fig1)

with tab2:
    fig2, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
    hrs = res["hours"]
    for i in range(res["n_houses"]):
        axes[0].plot(hrs, res["soc_hist"][i] * 100, color="steelblue", alpha=0.2)
    axes[0].plot(hrs, res["soc_hist"][0] * 100, color="#d90429", lw=2, label="H01 (Low Start: 22%)")
    axes[0].plot(hrs, res["soc_hist"][-1] * 100, color="#2b9348", lw=2, label=f"H{res['n_houses']:02d} (High Start: 85%)")
    axes[0].set_ylabel("Battery SoC (%)")
    axes[0].legend(loc="lower right")
    axes[0].grid(True, alpha=0.3)

    axes[1].axhline(0, color="black", lw=0.8)
    axes[1].plot(hrs, res["p_share_hist"][-1], color="#2b9348", lw=1.8, label=f"H{res['n_houses']:02d} (+Export/-Import)")
    axes[1].plot(hrs, res["p_share_hist"][0], color="#d90429", lw=1.8, label="H01 (+Export/-Import)")
    axes[1].set_ylabel("DC Bus Power (kW)")
    axes[1].legend(loc="upper right")
    axes[1].grid(True, alpha=0.3)

    axes[2].axhline(dc_bus_v, color="black", ls="--", lw=1)
    axes[2].fill_between(hrs, res["v_bus_hist"].min(axis=0), res["v_bus_hist"].max(axis=0), color="purple", alpha=0.2, label="Bus Voltage Spread")
    axes[2].plot(hrs, res["v_bus_hist"].mean(axis=0), color="purple", lw=2, label="Mean DC Voltage")
    axes[2].set_ylabel("DC Bus (V)")
    axes[2].set_xlabel("Hour of Day")
    axes[2].legend(loc="upper right")
    axes[2].grid(True, alpha=0.3)
    st.pyplot(fig2)

with tab3:
    col_a, col_b = st.columns(2)
    with col_a:
        fig3a, ax3a = plt.subplots(figsize=(6, 4.5))
        cb = res["cost_breakdown"]
        bars = ax3a.bar(list(cb.keys()), [v / 1e3 for v in cb.values()], color=["#2a9d8f", "#e9c46a", "#e76f51", "#264653"])
        ax3a.set_ylabel("CapEx ($ Thousands USD)")
        ax3a.set_title("SPV Hardware & Cabling CapEx Breakdown")
        for b in bars:
            ax3a.text(b.get_x() + b.get_width()/2, b.get_height() + 1, f"${b.get_height():.1f}k", ha="center", fontsize=8.5, fontweight="bold")
        st.pyplot(fig3a)
    with col_b:
        fig3b, ax3b = plt.subplots(figsize=(6, 4.5))
        yrs = np.arange(16)
        cum = np.cumsum(res["cashflows_m_kes"])
        ax3b.axhline(0, color="black", ls="--", lw=1)
        ax3b.axvline(lease_yrs, color="orange", ls=":", label=f"Transfer Yr {lease_yrs}")
        ax3b.plot(yrs, cum, marker="o", color="#1d3557", lw=2)
        ax3b.fill_between(yrs, 0, cum, where=(cum >= 0), color="#2b9348", alpha=0.2)
        ax3b.set_xlabel("Project Year")
        ax3b.set_ylabel("Cumulative Cash Flow (M KES)")
        ax3b.set_title(f"15-Yr SPV Cash Trajectory (IRR = {res['irr_pct']}%)")
        ax3b.legend()
        st.pyplot(fig3b)

with tab4:
    st.subheader("Combined SPV Portfolio Summary (Avana + The Kongo's Estate)")
    port_rows = []
    for name, path in preset_estates.items():
        if os.path.exists(path):
            r = run_estate_simulation(
                path, name.split(" (")[0], pv_kwp, batt_kwh, dc_bus_v, cable_mm2,
                kplc_tariff, discount_pct, fixed_fee, anchor_kwh, lease_yrs,
                pv_usd, batt_usd, router_usd
            )
            port_rows.append({
                "Estate": r["estate_name"],
                "Homes": r["n_houses"],
                "DC Ring (m)": r["ring_m"],
                "SPV CapEx ($)": r["capex_usd"],
                "CapEx/Home ($)": r["capex_per_hh_usd"],
                "Self-Sufficiency (%)": r["self_suff_pct"],
                "Payback (Yrs)": r["payback_yr"],
                "15-Yr IRR (%)": r["irr_pct"],
            })
    if port_rows:
        df_port = pd.DataFrame(port_rows)
        st.dataframe(df_port, use_container_width=True)
        tot_homes = df_port["Homes"].sum()
        tot_capex = df_port["SPV CapEx ($)"].sum()
        avg_irr = round(df_port["15-Yr IRR (%)"].mean(), 1)
        st.success(f"**Combined Portfolio:** {tot_homes} Homes | Total SPV CapEx: **${tot_capex:,}** | Average Unlevered IRR: **{avg_irr}%**")
```
